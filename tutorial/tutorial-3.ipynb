# Tutorial n°3

## Harmonic spectra of Coronene monomer and dimer

### A. Optimization

```python
import deMonPy

deMonPy.configure_from_file("../global.json")
```

```python
from deMonPy.molden import read_XYZ

images, comm = read_XYZ("./data/coronene.xyz")
coro = images[-1]
```

```python
from deMonPy.deMonNano import deMonNano

WORKDIR = ".run/tutorial-3/A/"

parameters = {
    "DEMON_EXECUTABLE": deMonPy.DEMON_EXECUTABLE,
    "BASIS": {"PTYPE": "BIO", "SKFILE": deMonPy.DEMON_BASIS},
    "DEMON_PARAMETERS": {
        "ACTIVE": {
            "DFTB": {"SCC": True},
        },
    },
    "DEMON_MODULE": {"ACTIVE": {"OPT": {"MAX": 9999, "OUT": 1, "TRAJECTORY": True}}},
}

dem = deMonNano(title="CALCULATION DEMONANO", workdir=WORKDIR, **parameters)

dem.calculate(symbols=coro.symbols, positions=coro.positions)
```

```python
from ase.visualize import view

view(dem.results["trajectory"])
```

### B. Frequencies analysis

```python
optimized_geometry = dem.results["output_geometry"]
print("Optimized geometry : ", optimized_geometry)
```

```python
from deMonPy.deMonNano import deMonNano

WORKDIR = ".run/tutorial-3/B/"

parameters = {
    "DEMON_EXECUTABLE": deMonPy.DEMON_EXECUTABLE,
    "BASIS": {"PTYPE": "BIO", "SKFILE": deMonPy.DEMON_BASIS},
    "DEMON_PARAMETERS": {
        "ACTIVE": {"DFTB": {"SCC": True}, "FREQ": True},
    },
}

dem = deMonNano(title="CALCULATION DEMONANO", workdir=WORKDIR, **parameters)

dem.calculate(symbols=coro.symbols, positions=coro.positions)
```

```python
import numpy as np

results = dem.results

freq = np.array([res["frequency"] for res in results["frequency"]])

print("Available frequencies in coronene : \n", freq)
```

### C. Dimer of coronene

```python
shift_coronene = optimized_geometry.copy()
shift_coronene.positions[:, 1] += 3.35
dimer = optimized_geometry + shift_coronene
```

```python
from deMonPy.deMonNano import deMonNano

WORKDIR = ".run/tutorial-3/C/"

parameters = {
    "DEMON_EXECUTABLE": deMonPy.DEMON_EXECUTABLE,
    "BASIS": {"PTYPE": "BIO", "SKFILE": deMonPy.DEMON_BASIS},
    "DEMON_PARAMETERS": {
        "ACTIVE": {
            "DFTB": {"SCC": True, "DISP": 2},
        },
    },
    "DEMON_MODULE": {"ACTIVE": {"OPT": {"MAX": 9999, "OUT": 1, "TRAJECTORY": True}}},
}

dem = deMonNano(title="CALCULATION DEMONANO", workdir=WORKDIR, **parameters)

dem.calculate(symbols=dimer.symbols, positions=dimer.positions)
```

```python
view(dem.results["trajectory"])

optimized_geometry = dem.results["output_geometry"]
print("Optimized geometry : ", optimized_geometry)
```

### E. Twisted dimer of coronene

```python
optimized_dimer.positions -= optimized_dimer.get_center_of_mass()

coro_1 = optimized_dimer[:36].copy()
coro_2 = optimized_dimer[36:].copy()

coro_2.rotate(30.0, "y")

twisted_dimer = coro_1 + coro_2
```

```python
parameters = {
    "DEMON_EXECUTABLE": deMonPy.DEMON_EXECUTABLE,
    "BASIS": {"PTYPE": "BIO", "SKFILE": deMonPy.DEMON_BASIS},
    "DEMON_PARAMETERS": {
        "ACTIVE": {"DFTB": {"SCC": True, "DISP": 2}},
    },
}
```

```python
from deMonPy.deMonNano import deMonNano

WORKDIR = ".run/tutorial-3/E/"

parameters.update(
    {"DEMON_MODULE": {"ACTIVE": {"OPT": {"MAX": 9999, "OUT": 1, "TRAJECTORY": True}}}}
)

dem = deMonNano(title="CALCULATION DEMONANO", workdir=WORKDIR, **parameters)

dem.calculate(symbols=twisted_dimer.symbols, positions=twisted_dimer.positions)
```

```python
optimized_dimer = dem.results["output_geometry"]
```
