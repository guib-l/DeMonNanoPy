## 1. How it works

This is the most important files to run a deMonNano calculation : 

<div style="background-color: #feede7; border-left: 6px solid #f33d21; padding: 10px; margin-bottom: 10px; margin-left:10px;">
<b>deMon.inp</b> : thes input file </br>
<b>deMon.out</b> : the main output file </br>
<b>deMon.mol</b> : final geometry of molecule, use `molden deMon.mol` to visualize the geometries
</div>

Th input file `deMon.inp` looks like : 
```txt 
MAINKEYWORD SUBKEY1 SUBKEY2 ...
MAINKEYWORD’ SUBKEY’1 SUBKEY’2 ...
```

The geometry should be given at the end like this
```txt 
SYMBOL_1 POSITION_X POSITION_Y POSITION_Z
SYMBOL_2 POSITION_X POSITION_Y POSITION_Z
SYMBOL_3 POSITION_X POSITION_Y POSITION_Z
...
```

Please, visite http://demon-nano.ups-tlse.fr/ for list of keywords

<div style="background-color: #fef4e7; border-left: 6px solid #f39c21; padding: 10px; margin-bottom: 10px; margin-left:10px;">
    <b>Note:</b></br>
Each line starting with <b>#</b> is not readen by the deMon input parser.
</div>

The output file `deMon.out` containt all detailed results and time analysis.

## 2. Start a new calculation

To create a new calculation, write a new file `deMon.inp` in somewhere : 
```txt 
DFTB
PARAM PTYPE=NSC
∼/basis
GEOMETRY
O  1.2478 -0.5185 3.4049
H  1.5946 -1.4204 3.3886
H  0.9008 -0.3341 2.5062
```
Run this calculation with the deMonNano binary provided in the website.

With this example, we ask a simple DFTB calculation. The `PARAM` keyword specifies the location of the basis. 

<div style="background-color: #fef4e7; border-left: 6px solid #f39c21; padding: 10px; margin-bottom: 10px; margin-left:10px;">
    <b>Note:</b></br>
It is strongly recommended that you change the basis location with your current basis directory
</div>

When you finish to explore the `deMon.out` file, try to start a new calculation with `DFTB SCC`. We need to change `NSC` by `BIO` or `MAT`.

Now, below the first line in your input file, add `OPTIMIZATION`. This key will specify to start an optimization of geometry. 
At the end of calculation, plot trajectory with your favourite visualization software. For the exemple, we used molden : `molden deMon.mol`.

<div style="background-color: #fef4e7; border-left: 6px solid #f39c21; padding: 10px; margin-bottom: 10px; margin-left:10px;padding-right:30px;">
    <b>Note:</b></br>
You found the optimized geometry at the end of deMon.mol. Details of the optimisation are recorded in deMon.out. </br>The default algorithm optimiser is the conjugate gradient method.
</div>

Perform a single point SCC calculation on the optimized water molecule. Add debug option to write many information with the new line `PRINT DEBUG`.

**In the output file, you could observe the Hamiltonian, Overlaps, Density matrix.**
