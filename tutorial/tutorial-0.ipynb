### C. First calculation

Now, setup your first calculation with DeMonNanoPy :

````python
import numpy as np

symbols = ["O","H","H","O","H","H"]
positions = np.array(
            [[1.2478,-0.5185,3.4049],
            [1.5946,-1.4204,3.3886],
            [0.9008,-0.3341,2.5062]]
        )
````

````python
import deMonPy
from deMonPy.deMonNano import deMonNano

WORKDIR = ".run/tutorial-0/"

parameters = {
        "DEMON_EXECUTABLE":EXECUTABLE,
        "BASIS":{
            "PTYPE":"NSC",
            "SKFILE":BASIS
        },
        "DEMON_PARAMETERS":{
            "ACTIVE":{
                "DFTB":{
                },
            },
        }
    }

dem = deMonNano(
        title="CALCULATION DEMONANO",
        workdir=WORKDIR,
        **parameters
    )

dem.calculate(
        symbols=symbols,
        positions=positions
    )

dem.print_results()
````
