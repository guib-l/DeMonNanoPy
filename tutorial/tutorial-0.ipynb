## Basics about deMonNano

### A. Base about DFTB model

The Density Functional based Tight-Binding model derived from Density Functional Theory model. The energy of DFTB is expressed as

$$
E_\mathrm{DFTB} = V^{rep} + \underset{i}{\sum^\mathrm{occ}} f_i\bra{\psi_i} H^0\ket{\psi_i} + \frac{1}{2} \underset{ab}{\sum}\Delta q_a\Delta q_b \gamma_\mathrm{ab}
$$

where the first term is strictly fitted repulsive terms as a sum of paire potential. In the second terms, $H^0$ is the hamiltonian operator and $f_i$ is occupations. Molecular orbitals are expressed in a minimal basis of valence atomic orbitals $\{\phi_\mu(\textbf{r})\}$, in a linear combination of atomic orbitals

$$
\ket{\psi_i} =  \underset{\mu }{\sum} c_\mu^i \ket{\phi_\mu}
$$

The third terms in energy expression is the second order contributions to the energy.

### C. First calculation

Now, setup your first calculation with DeMonNanoPy :

````python
import numpy as np

symbols = ["O", "H", "H", "O", "H", "H"]
positions = np.array(
    [[1.2478, -0.5185, 3.4049], [1.5946, -1.4204, 3.3886], [0.9008, -0.3341, 2.5062]]
)
````

````python
from deMonPy.deMonNano import deMonNano

WORKDIR = ".run/tutorial-0/"

parameters = {
    "DEMON_EXECUTABLE": deMonPy.DEMON_EXECUTABLE,
    "BASIS": {"PTYPE": "NSC", "SKFILE": deMonPy.DEMON_BASIS},
    "DEMON_PARAMETERS": {
        "ACTIVE": {
            "DFTB": {"SCC": False},
        },
    },
}

dem = deMonNano(title="CALCULATION DEMONANO", workdir=WORKDIR, **parameters)

dem.calculate(symbols=symbols, positions=positions)

dem.print_results()
````
