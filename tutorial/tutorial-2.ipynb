## Molecular dynamics of Coronene

### A. Simple dynamics

```python
import deMonPy

deMonPy.configure_from_file("../global.json")
```

```python
from deMonPy.molden import read_XYZ

images, comm = read_XYZ("./data/coronene.xyz")
coro = images[-1]
```

```python
from deMonPy.deMonNano import deMonNano

WORKDIR = ".run/tutorial-2/A/"

parameters = {
    "DEMON_EXECUTABLE": deMonPy.DEMON_EXECUTABLE,
    "BASIS": {"PTYPE": "BIO", "SKFILE": deMonPy.DEMON_BASIS},
    "DEMON_PARAMETERS": {
        "ACTIVE": {
            "DFTB": {"SCC": True},
        },
    },
    "DEMON_MODULE": {
        "ACTIVE": {
            "MD": {
                "MDYNAMICS": {
                    "ZERO": True,
                },
                "TIMESTEP": 0.5,
                "MDSTEP": {"MAX": 100, "OUT": 1},
                "TRAJECTORY": True,
            }
        }
    },
}

dem = deMonNano(title="CALCULATION DEMONANO", workdir=WORKDIR, **parameters)

dem.calculate(symbols=coro.symbols, positions=coro.positions)
```

Try to visualize molecular evolution in MD.

```python
from ase.visualize import view

view(dem.results["trajectory"])
```

### B. Dynamics at 300K

```python
from deMonPy.deMonNano import deMonNano

WORKDIR = ".run/tutorial-2/B/"

parameters = {
    "DEMON_EXECUTABLE": deMonPy.DEMON_EXECUTABLE,
    "BASIS": {"PTYPE": "BIO", "SKFILE": deMonPy.DEMON_BASIS},
    "DEMON_PARAMETERS": {
        "ACTIVE": {
            "DFTB": {"SCC": True},
        },
    },
    "DEMON_MODULE": {
        "ACTIVE": {
            "MD": {
                "MDYNAMICS": {
                    "RANDOM": 300,
                },
                "TIMESTEP": 0.5,
                "MDSTEP": {"MAX": 100, "OUT": 1},
                "TRAJECTORY": True,
            }
        }
    },
}

dem = deMonNano(title="CALCULATION DEMONANO", workdir=WORKDIR, **parameters)

dem.calculate(symbols=coro.symbols, positions=coro.positions)
```

```python
from ase.visualize import view

view(dem.results["trajectory"])
```
